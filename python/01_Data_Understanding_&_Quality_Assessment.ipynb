```python
def basic_info(df):
    summary = pd.DataFrame({
        'Data_Type': df.dtypes.astype(str),
        'Missing_Values': df.isnull().sum(),
        'Missing_Percentage': (df.isnull().mean() * 100).round(2),
        'Unique_Values': df.nunique()
    })

    summary.index.name = 'Column_Name'

    return summary.reset_index()
```

### 9.2 Duplicate Analysis

This section identifies duplicate records in each dataset. Both full-row duplicates and primary key duplicates are checked to ensure that records are not unintentionally repeated.

Duplicate records can affect analysis accuracy and lead to incorrect business insights.

```python
def duplicate_analysis(df, primary_key=None):

    full_duplicates = df.duplicated().sum()

    print(f"Total Rows: {len(df)}")
    print(f"Full Row Duplicates: {full_duplicates}")

    if primary_key:
        pk_duplicates = df[primary_key].duplicated().sum()

        print(f"Primary Key Duplicates: {pk_duplicates}")
```

### 9.5 Outlier Analysis

This section identifies potential outliers in numerical columns using the Interquartile Range (IQR) method.

The IQR method uses the first quartile (Q1), third quartile (Q3), and interquartile range to determine potential lower and upper outlier boundaries.

Potential outliers are investigated further to understand whether they represent unusual but valid business transactions or possible data quality issues.

# IQR Outlier Analysis

```python

def outlier_analysis(df, columns):

    results = []

    for col in columns:

        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)

        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        outliers = df[
            (df[col] < lower) | (df[col] > upper)
        ]

        results.append({
            'Column': col,
            'Lower_Bound': round(lower, 2),
            'Upper_Bound': round(upper, 2),
            'Outlier_Count': len(outliers),
            'Outlier_Percentage': round(
                len(outliers) / len(df) * 100, 2
            )
        })

    return pd.DataFrame(results)
```

## 9.8 Balance Reconciliation

```python
transactions_sorted = transactions.sort_values(
    by=['account_id', 'txn_date', 'transaction_id']
).copy()

display(transactions_sorted.head(10))
```

```python
transactions_sorted['previous_balance'] = (
    transactions_sorted.groupby('account_id')['balance_kes']
    .shift(1)
)

display(
    transactions_sorted[
        ['account_id', 'txn_date', 'txn_type',
         'amount_kes', 'previous_balance', 'balance_kes']
    ].head(10)
)
```

```python
transactions_sorted['expected_balance'] = np.where(
    transactions_sorted['txn_type'] == 'CREDIT',
    transactions_sorted['previous_balance'] + transactions_sorted['amount_kes'],
    transactions_sorted['previous_balance'] - transactions_sorted['amount_kes']
)

display(
    transactions_sorted[
        ['account_id', 'txn_type', 'amount_kes',
         'previous_balance', 'expected_balance', 'balance_kes']
    ].head(10)
)
```

```python
transactions_sorted['balance_difference'] = (
    transactions_sorted['balance_kes']
    - transactions_sorted['expected_balance']
)

display(
    transactions_sorted[
        ['account_id', 'txn_type', 'expected_balance',
         'balance_kes', 'balance_difference']
    ].head(10)
)
```

```python
balance_check = transactions_sorted[
    'balance_difference'
].dropna()

total_checked = len(balance_check)

mismatches = balance_check.ne(0).sum()

print("Total Transactions:", len(transactions_sorted))
print("Transactions Checked:", total_checked)
print("Balance Mismatches:", mismatches)
print("Matching Transactions:", (balance_check == 0).sum())
```

**Balance Reconciliation**

- Total transactions: 102,880
- Transactions checked: 101,980
- First transactions excluded: 900
- Balance mismatches: 0

**Conclusion:** All 101,980 checked transactions passed the balance reconciliation test. ## 9.11 Categorical Value Validation

```python
def categorical_validation(df, dataset_name, exclude_cols=None):

    if exclude_cols is None:
        exclude_cols = []

    categorical_cols = [
        col for col in df.select_dtypes(include='object').columns
        if col not in exclude_cols
    ]

    print(f"\n{'='*50}")
    print(f"Dataset: {dataset_name}")
    print(f"{'='*50}")

    for col in categorical_cols:

        print(f"\nColumn: {col}")

        print("\nValue Counts:")
        display(df[col].value_counts(dropna=False))

        print("\nUnique Values:")
        print(df[col].unique())

        extra_spaces = df[col].dropna().str.strip().ne(
            df[col].dropna()
        ).sum()

        print(f"\nLeading/Trailing Spaces: {extra_spaces}")
```
